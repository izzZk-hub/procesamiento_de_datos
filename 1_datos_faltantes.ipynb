# Datos Faltantes

```python
# Importar librerías necesarias
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.impute import SimpleImputer, KNNImputer
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cargar el conjunto de datos
diabetes = load_diabetes(as_frame=True)
data = diabetes.data
data['target'] = diabetes.target
```

```python
data.sample(10)
```

```python
# Simular valores faltantes
np.random.seed(42)
data.loc[data.sample(frac=0.1).index, 'bmi'] = np.nan
data.loc[data.sample(frac=0.1).index, 'bp'] = np.nan
```

### 2. Utiliza imputación simple (media) para llenar los valores faltantes de la columna 'bmi'.

```python
imputer = SimpleImputer(strategy='mean')

data['bmi'] = imputer.fit_transform(data[['bmi']]).ravel()

data['bmi'].isnull().sum()
```

### 3. Utiliza KNNImputer para imputar valores en las columnas 'bmi' y 'bp'. Compara los resultados con los de la imputación simple.

```python
diabetes = load_diabetes(as_frame=True)
data_knn = diabetes.data
data_knn['target'] = diabetes.target

np.random.seed(42)
data_knn.loc[data_knn.sample(frac=0.1).index, 'bmi'] = np.nan
data_knn.loc[data_knn.sample(frac=0.1).index, 'bp'] = np.nan

knn_imputer = KNNImputer(n_neighbors=5)

data_knn[['bmi', 'bp']] = knn_imputer.fit_transform(
    data_knn[['bmi', 'bp']]
)

data_knn[['bmi', 'bp']].isnull().sum()
```

### 4. Genera un histograma comparando los datos antes y después de la imputación en la columna 'bmi'.

```python
diabetes = load_diabetes(as_frame=True)
data_hist = diabetes.data

np.random.seed(42)
data_hist.loc[data_hist.sample(frac=0.1).index, 'bmi'] = np.nan

bmi_antes = data_hist['bmi'].copy()

imputer = SimpleImputer(strategy='mean')
data_hist['bmi'] = imputer.fit_transform(data_hist[['bmi']]).ravel()

plt.hist(bmi_antes.dropna(), bins=20, alpha=0.5, label='Antes')
plt.hist(data_hist['bmi'], bins=20, alpha=0.5, label='Después')

plt.xlabel('BMI')
plt.ylabel('Frecuencia')
plt.title('BMI antes y después de la imputación')
plt.legend()
plt.show()
```
